## Short options: cash-secured put
Use `tr_type='s'` for any short leg. A cash-secured put is a short put with cash set aside to buy the shares, so its expiry payoff is a short put's.
Charts are per share; multiply by 100 for one standard contract.

```python
csp = op.single_plotter(op_type='p', tr_type='s', spot=100, strike=95, op_pr=2.10, spot_range=15)
csp.update_layout(title_text='Cash-secured put  ·  Sell 95 put @ 2.10')
```

```python
s = op.summarize([op.Leg('p', 95, 's', 2.10)])
cash_reserved = 95 * 100                      # per contract
print(s)
print(f'Return if unassigned: {s.max_profit * 100 / cash_reserved:.2%} on ${cash_reserved:,} reserved')
```
